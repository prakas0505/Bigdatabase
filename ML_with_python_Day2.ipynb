```python
a = [1,2,3] # for loops
for i in range(100):
    print("Ram")
```

```python
# while loops
i = 0
while i<10:
    print("Ram")
    print(i)
    i = i + 1
```

```python
# control statements in loops

# break-- used to exit the loop without prematurely when a certain condition is met

#without break
for i in range(10):
  print(i)
```

```python
# with break
for i in range(10):
  if i == 5:
    break
  print(i)
```

```python
# continue statment -- skip current iteration
# without continue
for i in range(10):
  print(i)
```

```python
# with continue
for i in range(10):
  if i == 5:
    continue
  print(i)
```

```python
# conditional statements
x = 2

if x > 5:
    print("Big number")
else:
    print("Small number")
```
