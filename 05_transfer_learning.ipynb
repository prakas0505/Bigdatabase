# Tuning a Pretrained Model

This notebook is the **assignment brief** - it doesn't contain an
implementation itself. The worked solution lives in
`05_transfer_learning_keras.ipynb`, which:

1. loads a folder of labeled images,
2. takes a model pretrained on ImageNet (Xception),
3. replaces its final layer and trains only that new layer
   ("transfer learning"),
4. then unfreezes the whole model and trains a little more with a very
   small learning rate ("fine-tuning"),
5. and finally evaluates the result with a confusion matrix and example
   predictions.
